```python
# Encapsulation

# means hiding internal details of a class and only exposing what’s necessary. 
# It helps to protect important data from being changed directly and keeps the code secure and organized.
# Encapsulation is an OOP concept that binds data (variables) and methods (functions) together into a single unit (class) 
# and restricts direct access to the data to protect it from unauthorized modification.

# In simple words: Data hiding + data protection.

# Using access modifiers:

# Access Modifier	     Symbol	       Meaning
# Public	              name	    Accessible everywhere
# Protected	              _name	    Accessible within class & subclass
# Private	              __name	Accessible only inside the class

# exam short answer
#  Encapsulation is an OOP concept that wraps data and methods into a single unit and
# restricts direct access to data using access modifiers to provide data security.
```

```python
# single inheritance
# A child class inherits from just one parent class.

class Parent:
    def __init__(self, father_name):
        self.father_name = father_name  

class Child(Parent):
    def display(self, child_name):
        print(f"My name is {child_name}")               
        print(f"My father name is {self.father_name}")  


c1 = Child("Pratik")
c1.display("Rahul")

      
```

```python
# Multiple inheritance
# In multiple inheritance, a child class can inherit from more than one parent class.

class Father:
    def __init__(self, father_name):
        self.father_name = father_name

    def show_father(self):
        print(f"My father is {self.father_name}")


class Mother:
    def __init__(self, mother_name):
        self.mother_name = mother_name

    def show_mother(self):
        print(f"My mother is {self.mother_name}")

class Child(Father, Mother):
    def __init__(self, father_name, mother_name, child_name):
        Father.__init__(self, father_name)  
        Mother.__init__(self, mother_name)  
        self.child_name = child_name

    def show_child(self):
        print(f"My name is {self.child_name}")

c1 = Child("Pratik", "Anita", "Rahul")

c1.show_child()    
c1.show_father()   
c1.show_mother()   
```

```python
# multilevel inheritance
# In multilevel inheritance, a class is derived from another derived class (like a chain).

class Grandfather:
    def __init__(self, grandfather_name):
        self.grandfather_name = grandfather_name

    def show_grandfather(self):
        print(f"My grandfather is {self.grandfather_name}")

class Father(Grandfather):
    def __init__(self, grandfather_name, father_name):
        Grandfather.__init__(self, grandfather_name)  # call Grandfather constructor directly
        self.father_name = father_name

    def show_father(self):
        print(f"My father is {self.father_name}")


class Child(Father):
    def __init__(self, grandfather_name, father_name, child_name):
        Father.__init__(self, grandfather_name, father_name)  # call Father constructor directly
        self.child_name = child_name

    def show_child(self):
        print(f"My name is {self.child_name}")

c1 = Child("jayram", "Partik", "Rahul")

c1.show_child()        
c1.show_father()      
c1.show_grandfather()  
```

```python
# hierarchical inheritance
# In hierarchical inheritance, multiple child classes inherit from the same parent class.

class School:
    def __init__(self, school_name):
        self.school_name = school_name

    def show_school(self):
        print(f"School: {self.school_name}")

class Teacher(School):
    def __init__(self, school_name, teacher_name):
        School.__init__(self, school_name)  
        self.teacher_name = teacher_name

    def show_teacher(self):
        print(f"Teacher: {self.teacher_name}")

class Student(School):
    def __init__(self, school_name, student_name):
        School.__init__(self, school_name)  
        self.student_name = student_name

    def show_student(self):
        print(f"Student: {self.student_name}")

t1 = Teacher("bhavana high school", "Mr. Sharma")
s1 = Student("bhavana high school", "Ananya")

t1.show_school()    
t1.show_teacher()   

s1.show_school()    
s1.show_student()   
```

```python
# Hybrid inheritance
# Hybrid inheritance is a combination of more than one type of inheritance.

class Grandfather:
    def __init__(self, grandfather_name):
        self.grandfather_name = grandfather_name

    def show_grandfather(self):
        print(f"My grandfather is {self.grandfather_name}")


class Father(Grandfather):
    def __init__(self, grandfather_name, father_name):
        Grandfather.__init__(self, grandfather_name)
        self.father_name = father_name

    def show_father(self):
        print(f"My father is {self.father_name}")


class Mother:
    def __init__(self, mother_name):
        self.mother_name = mother_name

    def show_mother(self):
        print(f"My mother is {self.mother_name}")

class Child(Father, Mother):
    def __init__(self, grandfather_name, father_name, mother_name, child_name):
        Father.__init__(self, grandfather_name, father_name)  
        Mother.__init__(self, mother_name)                   
        self.child_name = child_name

    def show_child(self):
        print(f"My name is {self.child_name}")


c1 = Child("jayram", "Pratik", "Suman", "Rahul")

c1.show_child()
c1.show_father()
c1.show_mother()
c1.show_grandfather()
```

```python
# Polymorphism means one function name works in many forms.
# The same len() function behaves differently depending on the object type passed to it.
# This is called function polymorphism.
# len() is a built-in Python function used to find the number of elements in an object.

print(len("Python"))            
print(len([10, 20, 30, 40]))    
print(len((1, 2, 3)))           
print(len({1, 2, 3, 4}))        
print(len({"a": 1, "b": 2}))  
```

```python
# len() is polymorphic because its behavior depends on the runtime object and its implementation of the __len__() special method.
# len() is just a wrapper function
# Real logic lives inside __len__()

class Student:
    def __init__(self, name, subjects):
        self.name = name
        self.subjects = subjects

    def __len__(self):
        return len(self.subjects)

s1 = Student("Hinal", ["Python", "DBMS", "AI"])
s2 = Student("Amit", ["Math", "Physics"])

print(len(s1))
print(len(s2))
```

```python
# Task 7: Bank Account System Create base class 'BankAccount' with method withdraw(amount).
# Create child classes: SavingsAccount, CurrentAccount with different withdrawal rules.

class BankAccount:
    def __init__(self,balance):
        self.balance=balance
    def withdraw(self,amount):
        pass

class SavingsAccount(BankAccount):
    def __init__(self, balance):
        super().__init__(balance)
        self.min_balance = 500

    def withdraw(self, amount):
        if amount <= 0:
            return "Invalid amount"

        if amount > self.balance:
            return "Insufficient balance"

        if self.balance - amount < self.min_balance:
            return "Minimum balance rule violated"

        self.balance -= amount
        return f"Savings withdrawal successful. Balance: ₹{self.balance}"
    
class CurrentAccount(BankAccount):
    def __init__(self, balance):
        super().__init__(balance)
        self.min_balance = 500

    def withdraw(self, amount):
        if amount <= 0:
            return "Invalid amount"

        if amount > self.balance:
            return "Insufficient balance"

        if self.balance - amount < self.min_balance:
            return "Minimum balance rule violated"

        self.balance -= amount
        return f"Savings withdrawal successful. Balance: ₹{self.balance}"
    
bank =[SavingsAccount(5000),CurrentAccount(10000)]

for account in bank:
    # print(account.withdraw(11000))
    print(account.withdraw(7000))
    # print(account.withdraw(3000))
    


```

```python
# Abstraction
# Data abstraction means showing only the essential features and hiding the complex internal details.
# abstraction is used to hide the implementation details from the user and expose only necessary parts, making the code simpler and easier to interact with.

# an Abstract Base Class (ABC) is used to achieve data abstraction by defining a common interface for its subclasses. 
# It cannot be instantiated directly and serves as a blueprint for other classes.

# Abstract classes are created using abc module and @abstractmethod decorator, 
# allowing developers to enforce method implementation in subclasses while hiding complex internal logic.

from abc import ABC, abstractmethod

class Greet(ABC):
    @abstractmethod
    def say_hello(self):
        pass  # Abstract method

class English(Greet):
    def say_hello(self):
        return "Hello!"

g = English()
print(g.say_hello())
```

```python
"""Task 3:Employee Salary System ()
Objective: Create an abstract class Employee.
Requirements:
- Create abstract method calculate_salary().
- Create subclasses FullTimeEmployee and PartTimeEmployee.
- Calculate and display salary."""

from abc import ABC , abstractmethod

class Employee(ABC):
    @abstractmethod
    def calculate_salary(self,salary):
        pass


class FullTimeEmployee(Employee):
    def calculate_salary(self,salary):
        return f"Full time Employee salary is :{salary * 20 }"
    
class PartTimeEmployee(Employee):
    def calculate_salary(self, salary):
        return f"part time employee salary is :{salary * 10}"
    

f=FullTimeEmployee()
p=PartTimeEmployee()

print(f.calculate_salary(1000))
print(p.calculate_salary(1000))
```

```python
"""Task 4: Shape Area Calculator ()
Objective: Create an abstract class Shape.
Requirements:
- Create abstract method area().
- Create subclasses Circle, Rectangle, Triangle.
- Calculate and display area."""

from abc import ABC ,abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        return 0
        
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius

    def area(self):
      return f"Circle ares :{3.14 * self.radius * self.radius}"
    
class Rectangle(Shape):
    def __init__(self,length,width):
       self.length =length
       self.width = width
    
    def area(self):
        return f"Rectangle area :{self.length * self.width}"
    
class Triangle(Shape):
    def __init__(self,base,heigth):
        self.base=base
        self.heigth=heigth

    def area(self):
        return f"Triangle area :{0.5 * self.base * self.heigth}"
    
c = Circle(5)
r = Rectangle(10,5)
t= Triangle(4,5)

print("area of circle :",c.area())
print("area of Rectangle :",r.area())
print("area of Triangle :",t.area())
```
