Pandas Series 
A Pandas Series is a one-dimensional labeled array capable of holding data of any type such as integers, floats, strings, or objects.

It is similar to a single column in a table or spreadsheet.

In simple words:

Series = Data + Index (labels)

```python
# Series from List
import pandas as pd
data =[10,20,30,40,50]
s=pd.Series(data)
print(s)
```

```python
# Series with Custom Index
import pandas as pd
data=[10,20,30,40,50]
s=pd.Series(data,index=["a","b","c","d","e"])
print(s)
```

```python
# Series from Dictionary
import pandas as pd
data={
    "maths":60,
    "science":70,
    "gujrati":50
    }
s=pd.Series(data)
print(s)
```

```python
# Accessing Elements in Series
import pandas as pd
data={
    "maths":60,
    "science":70,
    "gujrati":50,
    "english":80,
    "hindi":55,
    "social science":68
    }
dic=pd.Series(data)
print(dic)
print(dic["gujrati"])
print(dic[0:4])

```

```python
# Series Attributes
import pandas as pd
data={
    "maths":60,
    "science":70,
    "gujrati":50,
    "english":80,
    "hindi":55,
    "social science":68
    }
dic=pd.Series(data)
print(dic)
# dic.values show the values
print(dic.values)
# .index show the index
print(dic.index)
# .dtype show the data type
print(dic.dtype)
# .shape show the  number of elements.
print(dic.shape)
```

```python
# Mathematical Operations on Series
import pandas as pd
s=pd.Series([10,20,30,40,50])
print(s + 5)
print(s * 2)
print(s - 5)
print(s / 2)
```

```python
# Conditional Selection
import pandas as pd
s=pd.Series([200,300,400,500,600])
print(s[s > 300])
```

```python
# Basic Statistical Functions
import pandas as pd
s=pd.Series([10,15,20,25,30,35,40,45,50])
print(s.mean())
print(s.max())
print(s.min())
```

```python
# Updating Values & Deleting values
s = pd.Series([10,20,30])
s[1] = 50
print(s)
s.drop(1)
```

Pandas DataFrame 

A Pandas DataFrame is a two-dimensional data structure used to store data in rows and columns, similar to an Excel spreadsheet, SQL table, or CSV file.

It is the most important data structure in Pandas and is widely used in data analysis and machine learning.

Simple Definition 

A DataFrame is a two-dimensional labeled data structure in Pandas that contains rows and columns and can store data of different types.

```python
# Create DataFrame from List
import pandas as pd
data=[[10,20,30],[40,50,60],[70,80,90]]
df=pd.DataFrame(data)
print(df)
```

```python
# Create DataFrame from Dictionary
import pandas as pd
data = {
    "Name": ["Ram", "Shyam", "Riya"],
    "Age": [20, 21, 19],
    "City": ["Delhi", "Mumbai", "Pune"]
}
df = pd.DataFrame(data)
print(df)
```

```python
# selecting the data
import pandas as pd
data = {
    "Name": ["Ram", "Shyam", "Riya"],
    "Age": [20, 21, 19],
    "City": ["Delhi", "Mumbai", "Pune"]
}
df = pd.DataFrame(data)
print(df)
print("____________________")
print(df["Name"])
print("____________________")
print(df[["Name","Age"]])
print("____________________")
print(df.iloc[0])
print("____________________")
print(df.loc[1])
```

```python
#Filtering Data
import pandas as pd
data = {
    "Name": ["Ram", "Shyam", "Riya"],
    "Age": [20, 21, 19],
    "City": ["Delhi", "Mumbai", "Pune"]
}
df = pd.DataFrame(data)
print(df)
print("____________________")
print(df[df["Age"] > 20])
print("____________________")
df["Salary"] = [20000,25000,22000]
print(df)
print("____________________")
df.loc[0,"Age"] = 25
print(df)
df.drop("Salary", axis=1)
```

```python
# shorting  data
import pandas as pd
data = {
    "Name": ["Ram", "Shyam", "Riya"],
    "Age": [20, 21, 19],
    "City": ["Delhi", "Mumbai", "Pune"]
}
df = pd.DataFrame(data)
print(df.sort_values("Age"))
print("____________________")
print(df.sort_values("Age", ascending=False))
```

```python
# interpolate() is used to fill missing values (NaN) by estimating values between existing data points. 
# It works best with numeric columns.
import pandas as pd

df = pd.read_csv("student_performance_updated_1000.csv")

# df = df.interpolate()
df["StudyHoursPerWeek"] = df["StudyHoursPerWeek"].interpolate()

print(df["StudyHoursPerWeek"])
```

```python
# The dropna() function used to removes rows or columns with NaN values. 
# It can be used to drop data based on different conditions.
import pandas as pd

df = pd.read_csv("student_performance_updated_1000.csv")
# Remove Rows with Missing Values
df = df.dropna()
print(df)
# Remove Rows if a Specific Column Has Missing Values
df = df.dropna(subset=["Gender"])
print(df)
# Remove Columns with Missing Values
df = df.dropna(axis=1)
# Drop Rows with All Values Missing
df = df.dropna(how="all")
# Drop Rows with Any Missing Values
df = df.dropna(how="any")
```

String Manipulations 

String manipulation refers to cleaning, transforming, and processing text data so it becomes suitable for analysis. Pandas provides a wide collection of .str functions that make it easy to work with string columns inside a DataFrame such as converting cases, trimming spaces, splitting, extracting patterns, replacing values, and more.

```python
import pandas as pd
df = pd.read_csv("student_performance_updated_1000.csv")
print(df)
print(df.astype('string'))


```

String Operations

```python
#  lower()
print(df["Name"].str.lower())
```

```python
# Upper()
print(df["Name"].str.upper())
```

```python
# strip():This method removes unwanted leading and trailing spaces from each string to clean the data.
print(df["Name"].str.strip())

```

```python
# split(): This method splits each string into a list of parts based on a given separator.

# " " it means that in between space in the name replace with the ","
print(df["Name"].str.split(" ")) 
# " e " it means that given chracter in the strinf split with ",".
print(df["Gender"].str.split("e"))
```

```python
# len(): This method calculates and returns the character length of each string in the column.

print(df["Name"].str.len())
```

```python
#  cat(): This method concatenates all strings in the column into a single string using a chosen separator
print(df["ParentalSupport"].str.cat(sep=","))
print(df["Name"].str.cat(sep="$"))
```

```python
#  get_dummies(): This method converts each unique string into a separate one-hot encoded column for modeling.
print(df["Gender"].str.get_dummies())
```

```python
# startswith(): This method checks whether each string begins with the specified prefix.
print(df["Name"].str.startswith("S"))
print(df["Name"].str.startswith("J"))
```

```python
# endswith(): This method checks whether each string ends with the specified suffix.
print(df['Name'].str.endswith("a"))
print(df['Name'].str.endswith("s"))
```

```python
# replace(): This method replaces occurrences of a specific substring or pattern with a new value.

print(df['Name'].str.replace('John', 'Emily'))
```

```python
# repeat(): This method duplicates each string a given number of times.
print(df['Name'].str.repeat(2))
```

```python
# count(): This method counts how many times a specific substring or pattern appears in each string.

print(df['Name'].str.count('a'))
```

```python
# find(): This method returns the index of the first occurrence of a pattern within each string.
print(df['Name'].str.find('a'))
```

```python
# findall(): This method returns a list of all occurrences of a pattern found in each string.
print(df['Name'].str.findall('a'))
```

```python
# islower(): This method checks whether all characters in each string are lowercase.
print(df['Name'].str.islower())
```

```python
# isupper(): This method checks whether all characters in each string are uppercase.
print(df['Name'].str.isupper())
```

```python
# isnumeric(): This method checks whether each string contains only numeric characters.
print(df["Name"].str.isnumeric())
```

```python
# swapcase(): This method swaps uppercase letters to lowercase and lowercase letters to uppercase for each string.
print(df["Name"].str.swapcase())
```

```python
import pandas as pd 
data = {'A': ['foo', 'bar', 'foo', 'bar'], 'B': [1, 2, 3, 4]}
df = pd.DataFrame(data)
result = df.groupby('A')['B'].transform('mean')
print(result)
```
